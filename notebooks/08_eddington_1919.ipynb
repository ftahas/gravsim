# Lesson 08 — Eddington 1919

**Phase 6 of the gravsim project: light.**

Everything so far was about slow bodies: Mercury at $v/c = 1.6\times10^{-4}$, the 1PN correction
at $10^{-8}$ of Newton's force. Light is the opposite limit, $v = c$ exactly, and it is where
General Relativity was first *seen*. In 1801 Johann von Soldner, treating light as Newtonian
corpuscles, computed that a ray grazing the Sun is bent by $0.87''$; in 1911 Einstein got the
same number from the equivalence principle alone; in 1915 the full theory doubled it to
$1.75''$. On 29 May 1919 Arthur Eddington (Príncipe) and Andrew Crommelin (Sobral) photographed
the Hyades around the eclipsed Sun, and in November Dyson, Eddington and Davidson announced
$1.98'' \pm 0.12''$ and $1.61'' \pm 0.30''$: Einstein's value, not Soldner's. The *Times* headline
read "Revolution in science".

This lesson integrates the photon's geodesic exactly (`gravsim.metric.null`), reproduces both
numbers and the factor two between them, draws the rays, and follows the same physics to the
black-hole shadow and to the fourth classical test, Shapiro's time delay.

By the end you should be able to

1. write the null Binet equation and explain why it has no Newtonian term;
2. derive $\delta = 4GM/(c^2b)$ and say where Soldner's factor $\tfrac12$ comes from (PPN $\gamma$);
3. compute the deflection for any impact parameter, from the solar limb to the photon sphere,
   and identify the critical impact parameter $b_{\rm crit} = 3\sqrt3\,GM/c^2$ with the shadow;
4. compute the Shapiro delay for an Earth–Mars signal at superior conjunction ($\sim250\,\mu$s
   round trip) and explain what Viking (1979) and Cassini (2003) measured.


## 1. The physics: light on a null geodesic

**The equation.** In the Schwarzschild metric of Lesson 05 a photon has $d\tau = 0$, so the
geodesic is parametrised by an affine parameter $\lambda$ and the normalisation of the
four-velocity is replaced by the null condition. The two Killing constants survive,
$E = (1 - r_s/r)\,c^2\,dt/d\lambda$ and $L = r^2\,d\phi/d\lambda$, and only their ratio matters:

$$ b = \frac{cL}{E}, $$

the **impact parameter**, the perpendicular distance of the incoming straight line from the
mass. With $u = 1/r$, $\mu = GM/c^2$ and $' = d/d\phi$ the radial equation becomes the
first integral $u'^2 = 1/b^2 - u^2 + 2\mu u^3$, whose derivative is the **null Binet equation**

$$ u'' + u = 3\mu\,u^2 . $$

Compare the timelike one, $u'' + u = GM/L^2 + 3\mu u^2$: the Newtonian term is gone — a photon
has no $GM/L^2$ scale — and the only term left is exactly the one that produced Mercury's
$43''$. Without it, $u = \sin\phi/b$ is the straight line $y = b$ ($\phi$ measured from the
incoming asymptote). In the dimensionless $w = bu$ the equation is $w'' + w = 3\epsilon w^2$ with
$\epsilon = \mu/b$: the whole problem depends on the geometry only through $\mu/b$, which for the
solar limb is $\mu/R_\odot = 2.1\times10^{-6}$.

**The deflection.** Perturbing in $\epsilon$ with $u(0) = 0$, $u'(0) = 1/b$ gives
$u = \sin\phi/b + (\mu/b^2)(2 - \sin^2\phi - 2\cos\phi)$, and $u = 0$ at $\phi = \pi + \delta$
yields

$$ \delta = \frac{4GM}{c^2 b} + \frac{15\pi}{4}\Big(\frac{GM}{c^2b}\Big)^2 + \dots $$

(Einstein 1915; the second term from Keeton & Petters 2005). At the limb, $1.7512''$.

**The factor two.** In the parametrised post-Newtonian metric,
$ds^2 = -(1 - 2U/c^2)c^2dt^2 + (1 + 2\gamma U/c^2)d\mathbf x^2$, a light ray sees the refractive
index $n = 1 + (1+\gamma)\,GM/(c^2\rho)$ and Fermat's principle bends it by

$$ \delta = \frac{1+\gamma}{2}\cdot\frac{4GM}{c^2b}. $$

The time part of the metric ($g_{tt}$: clocks run slow, the Newtonian potential) gives half;
the curvature of space ($g_{ij}$, weighted by $\gamma$) gives the other half. Soldner's
corpuscle on a Newtonian hyperbola, $\tan(\delta/2) = GM/(bv^2)$ with $v = c$, and Einstein's
1911 equivalence-principle argument both have $\gamma = 0$: $2GM/(c^2b) = 0.875''$. The 1919
expeditions measured $\gamma$; Mercury measures $(2 + 2\gamma - \beta)/3$ (Lesson 04).

**Capture.** The turning point $r_0$ solves $b^2 = r_0^2/(1 - r_s/r_0)$ and exists only for
$b > b_{\rm crit} = 3\sqrt3\,\mu$, the impact parameter of the photon sphere $r = 3\mu$. Rays
inside $b_{\rm crit}$ are captured; a black hole seen from a distance $D$ is a dark disc of
angular radius $b_{\rm crit}/D$, its **shadow**. Near $b_{\rm crit}$ the deflection diverges as
$-\ln(b/b_{\rm crit} - 1)$.

**The Shapiro delay.** Integrating $c\,dt = n\,dl$ along the chord between an emitter and a
receiver that passes the mass at distance $b$ gives an excess over the Euclidean light time of

$$ \Delta t = (1+\gamma)\frac{GM}{c^3}\ln\frac{r_e + r_r + R}{r_e + r_r - R}
\;\simeq\; (1+\gamma)\frac{GM}{c^3}\ln\frac{4r_er_r}{b^2} \quad (b \ll r), $$

with $R$ the chord (Shapiro 1964). $GM_\odot/c^3 = 4.9\,\mu$s, so the logarithm (12.5 for
Earth–Mars at the limb) is what makes the effect measurable.

**Units.** gravsim keeps $GM$ (AU$^3$/day$^2$) and $c$ (AU/day) explicit; angles are in radians
and converted to arcseconds, times to microseconds. Two settings again: the real Sun
($\mu = 1.48$ km, $R_\odot = 696\,000$ km) and geometric units $GM = c = 1$ ($\mu = 1$,
photon sphere at 3, $b_{\rm crit} = 5.196$) for the strong field.


```python
import math

import matplotlib.pyplot as plt
import numpy as np

from gravsim import constants as C
from gravsim.metric import null
from gravsim.metric import schwarzschild as sch
from gravsim.viz import PALETTE, figure_grid
from gravsim.viz.plots import INK, INK_SECONDARY

plt.rcParams["figure.dpi"] = 100
GM, c = C.GM_SUN, C.C_AU_DAY
MU = GM / c**2  # gravitational radius of the Sun [AU]
R_SUN = C.R_SUN_AU
ARCSEC = C.ARCSEC_PER_RAD
MICROSECOND = 1e-6 / C.DAY_S  # one microsecond in days
```

## 2. Soldner's $0.87''$ versus Einstein's $1.75''$

Three numbers for a grazing ray, all from $\mu/R_\odot$: Soldner/Einstein-1911 ($\gamma = 0$),
Einstein-1915 ($\gamma = 1$), and the exact null geodesic, which exceeds the weak-field value by
the second-order term $(15\pi/16)\,\mu/R_\odot \approx 6\times10^{-6}$ — far below anything the
1919 plates could see, and even below today's VLBI.


```python
print(f"gravitational radius of the Sun  mu = GM/c^2 = {MU:.3e} AU = {MU * C.AU_KM:.3f} km")
print(f"solar radius                     R_sun = {R_SUN:.5e} AU = {R_SUN * C.AU_KM:.0f} km,  mu/R_sun = {MU / R_SUN:.2e}")
print(f"critical impact parameter        b_crit = 3 sqrt(3) mu = {null.critical_impact_parameter(GM, c) * C.AU_KM:.2f} km")
print()
soldner = null.deflection_newtonian_corpuscle(R_SUN, GM, c) * ARCSEC
einstein = null.deflection_angle_weak(R_SUN, GM, c) * ARCSEC
exact = null.deflection_angle(R_SUN, GM, c) * ARCSEC
print(f"Soldner 1801 / Einstein 1911 (gamma = 0):  2 GM/(c^2 R_sun) = {soldner:.4f} arcsec")
print(f"Einstein 1915 (gamma = 1):                 4 GM/(c^2 R_sun) = {einstein:.4f} arcsec   (constants: {C.LIGHT_DEFLECTION_LIMB_ARCSEC})")
print(f"exact null geodesic:                                           {exact:.6f} arcsec   "
      f"(exact/weak - 1 = {exact / einstein - 1:.2e}; 15 pi/16 mu/R_sun = {15 * math.pi / 16 * MU / R_SUN:.2e})")
print()
print(f"{'gamma':>6s} {'(1+gamma)/2':>12s} {'limb deflection':>16s}")
for gamma in (0.0, 0.5, 1.0, 1.5):
    print(f"{gamma:6.1f} {(1 + gamma) / 2:12.2f} {null.solar_limb_deflection_arcsec(gamma):13.4f}\"")
```

## 3. The deflection curve and the measurements

`deflection_angle` integrates the null Binet equation from the incoming asymptote until the
ray escapes and returns $\phi_{\rm end} - \pi$ (DOP853, `rtol = 1e-12`; it integrates the
deviation from the straight line so the $10^{-5}$ rad answer is not lost in the subtraction).
Below, the exact deflection for $b$ from the limb to $100\,R_\odot$ against the $4GM/(c^2b)$ and
$2GM/(c^2b)$ lines, with the measurements scaled to the limb by the $1/b$ law — which is how the
eclipse results were quoted: the stars on the 1919 plates sat at $2$–$6\,R_\odot$ (shaded band),
where the shifts are only $0.3''$–$0.9''$. The right panel shows the second-order term,
$(15\pi/16)\,\mu/b$ relative; it is drawn only out to $30\,R_\odot$, beyond which the residual of
a $10^{-7}$ rad deflection reaches the integrator's $10^{-12}$ tolerance.

| year | experiment | limb value / $\gamma$ |
|---|---|---|
| 1919 | Sobral 4-inch / Príncipe (Dyson, Eddington & Davidson) | $1.98 \pm 0.12$ / $1.61 \pm 0.30$ (probable errors) |
| 1922 | Wallal, Lick (Campbell & Trumpler) | $1.72 \pm 0.11$ |
| 1973 | Mauritania (Texas) | $1.66 \pm 0.18$ |
| 1976 | radio interferometry, 3C279 (Fomalont & Sramek) | $(1+\gamma)/2 = 1.007 \pm 0.009$ |
| 2004 | VLBI 1979–1999 (Shapiro et al.) | $\gamma = 0.9998 \pm 0.0004$ |
| 2003 | Cassini Shapiro delay (Bertotti, Iess & Tortora) | $\gamma - 1 = (2.1 \pm 2.3)\times10^{-5}$ |


```python
MEASUREMENTS = (  # (label, limb-normalised deflection [arcsec], uncertainty)
    ("1919 Sobral", 1.98, 0.12),
    ("1919 Principe", 1.61, 0.30),
    ("1922 Wallal", 1.72, 0.11),
    ("1973 Mauritania", 1.66, 0.18),
    ("1976 radio", 1.007 * einstein, 0.009 * einstein),
    ("2004 VLBI", (1 + 0.9998) / 2 * einstein, 0.0002 * einstein),
)
b_grid = np.geomspace(1.0, 100.0, 60) * R_SUN
delta_exact = np.array([null.deflection_angle(b, GM, c) for b in b_grid]) * ARCSEC
delta_weak = null.deflection_angle_weak(b_grid, GM, c) * ARCSEC
delta_soldner = null.deflection_newtonian_corpuscle(b_grid, GM, c) * ARCSEC

fig, axes = figure_grid(2, panel=(5.6, 4.0))
ax = axes[0]
ax.plot(b_grid / R_SUN, delta_exact, color=PALETTE[0], lw=2.0, label="exact null geodesic")
ax.plot(b_grid / R_SUN, delta_weak, color=INK, lw=0.9, ls="--", label="4GM/(c²b)  (Einstein 1915)")
ax.plot(b_grid / R_SUN, delta_soldner, color=PALETTE[1], lw=0.9, ls="--", label="2GM/(c²b)  (Soldner 1801)")
ax.axvspan(2.0, 6.0, color="0.92", label="stars on the 1919 plates")
for k, (label, val, err) in enumerate(MEASUREMENTS):
    offset = 1.0 + 0.03 * k  # spread the limb points slightly in b so the bars are visible
    ax.errorbar(offset, val, yerr=err, fmt="o", ms=4, color=PALETTE[2 + k % 6], capsize=3, label=label)
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel("impact parameter b [R☉]"); ax.set_ylabel("deflection [arcsec]")
ax.set_title("light deflection by the Sun"); ax.legend(fontsize=7, loc="lower left"); ax.grid(True, lw=0.3)

ax = axes[1]
near = b_grid <= 30.0 * R_SUN  # beyond that the 1e-12 integration tolerance shows in the residual
ax.plot(b_grid[near] / R_SUN, delta_exact[near] / delta_weak[near] - 1.0, color=PALETTE[0], lw=1.5, label="exact / weak − 1")
ax.plot(b_grid[near] / R_SUN, 15 * math.pi / 16 * MU / b_grid[near], color=INK, lw=0.9, ls="--", label="(15π/16) μ/b")
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel("b [R☉]"); ax.set_ylabel("relative second-order term")
ax.set_title("how weak the weak field is"); ax.legend(fontsize=8); ax.grid(True, lw=0.3)
print(f"deflection at b = 2 R_sun: {np.interp(2.0, b_grid / R_SUN, delta_exact):.3f} arcsec, "
      f"at 6 R_sun: {np.interp(6.0, b_grid / R_SUN, delta_exact):.3f} arcsec")
```

## 4. A bundle of rays

`ray_bundle` returns the Cartesian trace of each ray, entering from the left along $y = b$ with
the mass at the origin. For the Sun a deflection of $1.75''$ ($8.5\times10^{-6}$ rad) is invisible on
any honest drawing, so the figure uses the project's two tricks:

* **a compact mass** in geometric units ($GM = c = 1$): the same equation with $b$ of a few
  $\mu$ instead of $10^6\,\mu$; the horizon ($r = 2$) and photon sphere ($r = 3$) are drawn;
* **the real Sun with the deviation from the straight line amplified** by $10^5$, i.e. we draw
  $y_{\rm amp} = b + k\,(y - b)$ — the same device as the amplified GR − Newton displacement of
  Lesson 02's app lab. The middle panel is the honest picture: straight lines.

Note which way the rays bend (towards the mass) and that the bending is concentrated near the
closest approach: a lens, but one whose strength falls as $1/b$ instead of rising with $b$.


```python
fig, axes = figure_grid(3, panel=(4.8, 4.2))

# (a) compact mass, geometric units
ax = axes[0]
b_geo = np.array([-30.0, -20.0, -14.0, -10.0, -8.0, -6.5, 6.5, 8.0, 10.0, 14.0, 20.0, 30.0])
for b, (x, y) in zip(b_geo, null.ray_bundle(b_geo, 1.0, 1.0, n_points=800, r_max=60.0)):
    ax.plot(x, y, color=PALETTE[0], lw=0.9)
    ax.axhline(b, color="0.85", lw=0.5, zorder=0)
th = np.linspace(0.0, 2.0 * math.pi, 200)
ax.fill(2.0 * np.cos(th), 2.0 * np.sin(th), color=INK, label="horizon r = 2")
ax.plot(3.0 * np.cos(th), 3.0 * np.sin(th), color=INK_SECONDARY, lw=0.8, ls="--", label="photon sphere r = 3")
ax.set_xlim(-40.0, 40.0); ax.set_ylim(-40.0, 40.0); ax.set_aspect("equal")
ax.set_xlabel("x [GM/c²]"); ax.set_ylabel("y [GM/c²]"); ax.set_title("compact mass, b = 6.5 … 30 μ")
ax.legend(fontsize=7, loc="lower left")

# (b) and (c) the Sun, honest and amplified
b_sun = np.array([-5.0, -3.0, -2.0, -1.5, -1.0, 1.0, 1.5, 2.0, 3.0, 5.0]) * R_SUN
window = 15.0 * R_SUN
bundle = null.ray_bundle(b_sun, GM, c, n_points=1500, r_max=1.0)
for ax, amp in ((axes[1], 1.0), (axes[2], 1e5)):
    for b, (x, y) in zip(b_sun, bundle):
        ax.plot(x / R_SUN, (b + amp * (y - b)) / R_SUN, color=PALETTE[0], lw=0.9)
    ax.fill(np.cos(th), np.sin(th), color=PALETTE[3], label="Sun")
    ax.set_xlim(-window / R_SUN, window / R_SUN); ax.set_ylim(-window / R_SUN, window / R_SUN)
    ax.set_aspect("equal"); ax.set_xlabel("x [R☉]"); ax.set_ylabel("y [R☉]")
    ax.set_title("the Sun, to scale" if amp == 1.0 else "the Sun, deviation × 10⁵")
axes[1].legend(fontsize=7, loc="lower left")
for b in b_sun[b_sun > 0]:
    print(f"b = {b / R_SUN:3.1f} R_sun: deflection {null.deflection_angle(b, GM, c) * ARCSEC:.3f} arcsec, "
          f"closest approach b - r_0 = {(b - null.closest_approach_radius(b, GM, c)) * C.AU_KM:.3f} km (= mu)")
```

## 5. The photon sphere, capture and the shadow

Geometric units. As $b$ decreases towards $b_{\rm crit} = 3\sqrt3 = 5.196$ the closest approach
$r_0$ drops towards the photon sphere $r = 3$ and the deflection grows past $90°$, $180°$, a
full loop: the ray circles the unstable null orbit before escaping, and the number of loops
grows as $-\ln(b/b_{\rm crit} - 1)$ (Darwin 1959; Bozza 2002: $\delta \simeq -\ln(b/b_{\rm crit}-1) - 0.40$).
Below $b_{\rm crit}$ there is no turning point: the ray is captured. The dark disc of radius
$b_{\rm crit}$ projected on the sky is the shadow imaged by the Event Horizon Telescope — rimmed
by the light that looped once (the "photon ring").


```python
b_crit = null.critical_impact_parameter(1.0, 1.0)
f_grid = np.geomspace(1.0001, 1000.0, 80)  # b / b_crit
delta_geo = np.array([null.deflection_angle(f * b_crit, 1.0, 1.0) for f in f_grid])
eps = 1.0 / (f_grid * b_crit)  # mu / b

fig, axes = figure_grid(2, panel=(5.6, 4.2))
ax = axes[0]
ax.plot(f_grid - 1.0, np.degrees(delta_geo), color=PALETTE[0], lw=2.0, label="exact")
ax.plot(f_grid - 1.0, np.degrees(4 * eps), color=INK, lw=0.9, ls="--", label="4 μ/b")
ax.plot(f_grid - 1.0, np.degrees(4 * eps + 15 * math.pi / 4 * eps**2 + 128 / 3 * eps**3), color=PALETTE[1], lw=0.9, ls="--", label="series to (μ/b)³")
ax.plot(f_grid - 1.0, np.degrees(-np.log(f_grid - 1.0) - 0.4002), color=PALETTE[2], lw=0.9, ls=":", label="−ln(b/b_crit − 1) − 0.40")
for n in (1, 2):
    ax.axhline(360.0 * n, color="0.8", lw=0.6)
ax.set_xscale("log"); ax.set_yscale("log"); ax.set_ylim(1e-2, 1e3)
ax.set_xlabel("b / b_crit − 1"); ax.set_ylabel("deflection [deg]")
ax.set_title("from the weak field to the photon sphere"); ax.legend(fontsize=8, loc="upper right"); ax.grid(True, lw=0.3)

ax = axes[1]
for k, f in enumerate((2.0, 1.2, 1.05, 1.005)):
    x, y = null.ray_path(f * b_crit, 1.0, 1.0, n_points=3000, r_max=25.0)
    ax.plot(x, y, color=PALETTE[k], lw=1.0, label=f"b = {f:.3f} b_crit: {np.degrees(null.deflection_angle(f * b_crit, 1.0, 1.0)):.0f}°")
ax.fill(2.0 * np.cos(th), 2.0 * np.sin(th), color=INK)
ax.plot(3.0 * np.cos(th), 3.0 * np.sin(th), color=INK_SECONDARY, lw=0.8, ls="--")
ax.set_xlim(-14.0, 14.0); ax.set_ylim(-14.0, 14.0); ax.set_aspect("equal")
ax.set_xlabel("x [GM/c²]"); ax.set_ylabel("y [GM/c²]"); ax.set_title("winding around the photon sphere")
ax.legend(fontsize=7, loc="lower left")

pc_au = C.ARCSEC_PER_RAD  # 1 parsec = 206265 AU
for name, mass_msun, dist_pc, eht in (("M87*", 6.5e9, 16.8e6, "42 ± 3"), ("Sgr A*", 4.0e6, 8.0e3, "52 ± 2")):
    diameter = 2 * null.critical_impact_parameter(mass_msun * GM, c) / (dist_pc * pc_au)
    print(f"{name:7s}: shadow diameter 2 b_crit/D = {diameter * ARCSEC * 1e6:5.1f} micro-arcsec  (EHT ring: {eht})")
print(f"the Sun : b_crit = {null.critical_impact_parameter(GM, c) * C.AU_KM:.1f} km; it would have to fit inside "
      f"r = 3 mu = {sch.photon_sphere_radius(GM, c) * C.AU_KM:.1f} km to cast a shadow")
```

## 6. The Shapiro delay: Viking 1979, Cassini 2003

Irwin Shapiro proposed in 1964 to bounce radar off Venus or Mercury near superior conjunction
and time the echo: the $(1+\gamma)GM/c^3\ln(4r_er_r/b^2)$ excess is $\sim200\,\mu$s on a round trip
of twenty minutes, and it has a tell-tale shape, a logarithmic spike as the planet passes behind
the Sun. The 1967 Haystack echoes confirmed it to 20 %. The Viking landers on Mars (1976–77,
Reasenberg et al. 1979) turned the radar into a transponder and reached $\gamma = 1.000 \pm 0.002$;
Cassini (2002, Bertotti, Iess & Tortora 2003) measured not the delay but its time derivative — the
Doppler shift $y = -d\Delta t/dt$, a few parts in $10^{10}$ at $b = 1.6\,R_\odot$ — and reached
$\gamma - 1 = (2.1 \pm 2.3)\times10^{-5}$, still the best value of $\gamma$.

Below: the one-way delay for Earth–Mars, the round trip, the conjunction form, and the exact
geodesic (`shapiro_delay_numeric`, which integrates $dt/d\phi$ along the ray and subtracts the
chord measured in isotropic coordinates); then the spike, with Earth and Mars on circular
coplanar orbits so that Mars passes exactly behind the Sun (the grey gap is the occultation).


```python
R_EARTH, R_MARS = 1.0, 1.5237
one_way = null.shapiro_delay(R_EARTH, R_MARS, R_SUN, GM, c)
conj = null.shapiro_delay_conjunction(R_EARTH, R_MARS, R_SUN, GM, c)
exact_delay = null.shapiro_delay_numeric(R_EARTH, R_MARS, R_SUN, GM, c)
light_time = (math.sqrt(R_EARTH**2 - R_SUN**2) + math.sqrt(R_MARS**2 - R_SUN**2)) / c
print(f"GM_sun / c^3 = {GM / c**3 / MICROSECOND:.3f} microseconds;  ln(4 r_e r_r / b^2) = {math.log(4 * R_EARTH * R_MARS / R_SUN**2):.2f}")
print(f"Earth-Mars at conjunction, b = R_sun:  one way {one_way / MICROSECOND:.2f} us, round trip {2 * one_way / MICROSECOND:.1f} us "
      f"(Viking: ~250 us) on a one-way light time of {light_time * C.DAY_S / 60:.1f} min  ->  {one_way / light_time:.1e} relative")
print(f"conjunction form {conj / MICROSECOND:.2f} us;  exact geodesic {exact_delay / MICROSECOND:.2f} us "
      f"(exact/weak - 1 = {exact_delay / one_way - 1:+.1e}, the second-order end-point and Fermat terms)")
cassini = null.shapiro_delay(R_EARTH, 8.43, 1.6 * R_SUN, GM, c)
print(f"Cassini 2002 (Saturn at 8.43 AU, b = 1.6 R_sun): one way {cassini / MICROSECOND:.1f} us; "
      f"gamma = 0 would give {null.shapiro_delay(R_EARTH, 8.43, 1.6 * R_SUN, GM, c, gamma=0.0) / MICROSECOND:.1f} us")

# the spike: circular coplanar orbits, Kepler periods, Mars fixed on the far side of the Sun
days = np.linspace(-120.0, 120.0, 2401)
rate = 2 * math.pi * (1 / (2 * math.pi * math.sqrt(R_EARTH**3 / GM)) - 1 / (2 * math.pi * math.sqrt(R_MARS**3 / GM)))
theta = rate * days  # Earth's angle from the anti-Mars direction
r_e = np.stack([R_EARTH * np.cos(theta), R_EARTH * np.sin(theta)], axis=1)
r_m = np.array([-R_MARS, 0.0])
chord = r_m - r_e
b_t = np.abs(r_e[:, 0] * chord[:, 1] - r_e[:, 1] * chord[:, 0]) / np.linalg.norm(chord, axis=1)
visible = b_t > R_SUN
delay_t = np.full(days.shape, np.nan)
delay_t[visible] = [null.shapiro_delay(R_EARTH, R_MARS, b, GM, c) for b in b_t[visible]]
check_days = np.array([-60.0, -20.0, -5.0, 5.0, 20.0, 60.0])
check = [null.shapiro_delay_numeric(R_EARTH, R_MARS, float(np.interp(d, days, b_t)), GM, c) for d in check_days]

fig, axes = figure_grid(2, panel=(5.6, 4.0))
ax = axes[0]
ax.plot(days, 2 * delay_t / MICROSECOND, color=PALETTE[0], lw=1.5, label="round trip, weak field")
ax.plot(check_days, 2 * np.array(check) / MICROSECOND, "o", color=INK, ms=4, label="exact geodesic")
ax.axvspan(days[~visible].min(), days[~visible].max(), color="0.85", label="Mars behind the Sun")
ax.set_xlabel("days from superior conjunction"); ax.set_ylabel("Shapiro delay [μs]")
ax.set_title("Earth–Mars radar echo (coplanar model)"); ax.legend(fontsize=8); ax.grid(True, lw=0.3)
ax = axes[1]
ax.plot(days, b_t / R_SUN, color=PALETTE[1], lw=1.2)
ax.axhline(1.0, color=PALETTE[3], lw=0.8, label="solar limb")
ax.set_yscale("log"); ax.set_xlabel("days from superior conjunction"); ax.set_ylabel("closest approach of the chord [R☉]")
ax.set_title("the geometry behind the spike"); ax.legend(fontsize=8); ax.grid(True, lw=0.3)
print(f"the chord recedes from the Sun at {np.gradient(b_t, days)[-1] * C.AU_KM / C.DAY_S:.1f} km/s; "
      f"at b = 10 R_sun (day {np.interp(10.0, b_t[days > 0] / R_SUN, days[days > 0]):.1f}) the round trip is still {2 * np.interp(10.0, b_t[days > 0] / R_SUN, delay_t[days > 0]) / MICROSECOND:.0f} us")
```

## 7. What the light tests measure: $\gamma$

Deflection and delay both scale with $(1+\gamma)$; Mercury's advance with $(2+2\gamma-\beta)/3$.
The perihelion alone cannot tell a theory with $\gamma = \tfrac12, \beta = 0$ (or any
$\beta = 2\gamma - 1$) from GR, but the light tests fix $\gamma$ first and the perihelion then
fixes $\beta$. With Cassini's
$\gamma$ and the planetary ephemerides, $|\beta - 1| < 10^{-4}$. The table runs the three
observables through gravsim's PPN hooks (`gamma=` in `deflection_angle_weak`, `shapiro_delay`,
and `gr_precession_arcsec_per_century` from Lesson 04).


```python
from gravsim.analysis import gr_precession_arcsec_per_century

A_MERCURY, E_MERCURY = 0.38710, 0.20564
print(f"{'gamma':>5s} {'beta':>5s} {'limb deflection':>16s} {'Earth-Mars delay':>17s} {'Mercury advance':>16s}")
for gamma, beta in ((1.0, 1.0), (0.0, 1.0), (0.5, 1.0), (0.5, 0.0), (1.0, 0.0)):
    defl = null.deflection_angle_weak(R_SUN, GM, c, gamma) * ARCSEC
    delay = null.shapiro_delay(R_EARTH, R_MARS, R_SUN, GM, c, gamma) / MICROSECOND
    prec = gr_precession_arcsec_per_century(A_MERCURY, E_MERCURY, GM, c, beta=beta, gamma=gamma)
    print(f"{gamma:5.1f} {beta:5.1f} {defl:13.3f}\"   {delay:13.1f} us   {prec:11.2f} \"/cy")
print()
print("measured: gamma - 1 = (2.1 ± 2.3)e-5 (Cassini 2003), (1+gamma)/2 = 0.9999 ± 0.0002 (VLBI 2004); "
      "Mercury 42.98 \"/cy (Lesson 04)")
```

## 8. Summary

* A photon obeys $u'' + u = 3\mu u^2$ — the timelike Binet equation without its Newtonian
  term; only $\mu/b$ matters.
* $\delta = \tfrac{1+\gamma}{2}\,4GM/(c^2b)$: $1.751''$ at the limb for GR, half of it for a
  Newtonian corpuscle or the equivalence principle alone, because they miss the curvature of
  space. 1919 chose GR; VLBI has since pinned $\gamma$ to $4\times10^{-4}$.
* The exact deflection exceeds $4\mu/b$ by $(15\pi/16)\mu/b$, diverges logarithmically at
  $b_{\rm crit} = 3\sqrt3\,\mu$ and is undefined inside: capture, the black-hole shadow.
* The Shapiro delay, $(1+\gamma)(GM/c^3)\ln(4r_er_r/b^2)$, is $124\,\mu$s one way for Earth–Mars
  at the limb; Viking measured it to $0.1\,\%$, Cassini (as a Doppler shift) to $2\times10^{-5}$.
