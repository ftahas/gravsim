# Lesson 06 — Le Verrier, Vulcan, Einstein

**Phase 5 of the gravsim project, part I: everybody curves spacetime.**

Lesson 04 gave the planets the relativistic field of the Sun and found the $43''$. This lesson
tells the story those 43 arcseconds have, replaces the Sun-only correction by the equations the
ephemerides actually integrate — Einstein, Infeld and Hoffmann's, in which *every* body curves
spacetime — and then reproduces the anomaly the way Le Verrier met it: not as a number we put
in, but as a residual between the best Newtonian model and the "observations". Our observations
are the JPL Horizons ephemeris, which contains General Relativity.

By the end you should be able to

1. tell the history: what Le Verrier measured in 1859, why a planet Vulcan was a reasonable
   idea, and what Einstein computed in November 1915;
2. say in words what the three groups of terms of the EIH equations do, and why JPL, INPOP
   and EPM integrate them rather than the test-particle term;
3. produce the anomaly *from data*: Mercury's perihelion in a Newtonian run drifts away from
   Horizons at $43''$ per century, and the EIH run does not;
4. explain why, for Mercury's secular rate, EIH and the test-particle 1PN term differ by only
   $\sim10^{-5}\,''$/cy — and show a system where the difference is large.

## 1. The physics: from a residual to the field equations

### 1.1 The story

**1859.** Urbain Le Verrier, who in 1846 had located Neptune from the misbehaviour of Uranus,
turned the same method on Mercury: 150 years of transits across the Sun, a complete Newtonian
theory of the planetary perturbations (the planets turn Mercury's ellipse by $\sim530''$ per
century, Lesson 02), and a residual. The observed advance of the perihelion exceeded the
computed one by $38''$ per century — Simon Newcomb's 1882 revision made it $43''$. Everything else
in the solar system fitted; this did not.

**Vulcan.** The method that had found Neptune suggested the cure: an unseen mass inside Mercury's
orbit, a planet or a ring of asteroids, which Le Verrier named Vulcan. Within months a country
doctor, Lescarbault, reported a transit; Le Verrier computed an orbit; for forty years amateurs
and professionals searched for Vulcan during total eclipses (Watson and Swift both "saw" it in
1878). Nothing was ever confirmed, and a body massive enough to move Mercury's perihelion would
have moved Venus too. Alternatives were tried — an oblate Sun, a $1/r^{2.00000016}$ force law
(Hall 1894), zodiacal dust — each creating a new problem elsewhere.

**November 1915.** In the week of 18 November Einstein applied his almost-finished field
equations to the motion of a planet in the field of the Sun, expanding to first order beyond
Newton, and found an advance of $43''$ per century for Mercury (and a halved-then-doubled light
deflection of $1.7''$, Lesson 08). No free parameter: the number came out of the theory. To
Ehrenfest he wrote in January 1916 that he had been "beside myself with joyous excitement" for
days; to Fokker, that he had had heart palpitations. Today's inertial-frame budget is
$531.6''$ (planets) $+ 42.98''$ (GR) $+ 0.03''$ (solar $J_2$) $= 574.6''$ against $574.1\pm0.4''$ observed.

### 1.2 The equations, in words

Einstein's 1915 calculation treated Mercury as a *test* body in the *static* field of the Sun —
Lesson 04's `make_pn1`. In 1938 Einstein, Infeld and Hoffmann derived from the field equations
the first post-Newtonian equations of motion of $N$ bodies of comparable masses. With
$\mu_j = GM_j$, $\mathbf r_{ij} = \mathbf r_i - \mathbf r_j$, barycentric velocities $\mathbf v$
and the Newtonian accelerations $\mathbf a^N$, in the PPN-parametrised form used by the
ephemerides (Moyer 2003, eq. 4-26; Will 2014, §3.4; derivation in `docs/theory/eih.md`):

$$
\begin{aligned}
\mathbf a_i = {}& \sum_{j\neq i}\frac{\mu_j\,\mathbf r_{ji}}{r_{ij}^3}\Big\{1
 - \frac{2(\beta+\gamma)}{c^2}\sum_{l\neq i}\frac{\mu_l}{r_{il}}
 - \frac{2\beta-1}{c^2}\sum_{k\neq j}\frac{\mu_k}{r_{jk}}
 + \gamma\frac{v_i^2}{c^2} + (1+\gamma)\frac{v_j^2}{c^2}
 - \frac{2(1+\gamma)}{c^2}\,\mathbf v_i\!\cdot\!\mathbf v_j
 - \frac{3}{2c^2}\Big(\frac{\mathbf r_{ij}\!\cdot\!\mathbf v_j}{r_{ij}}\Big)^2
 + \frac{1}{2c^2}\,\mathbf r_{ji}\!\cdot\!\mathbf a^N_j\Big\} \\
 &+ \frac{1}{c^2}\sum_{j\neq i}\frac{\mu_j}{r_{ij}^3}\Big[\mathbf r_{ij}\cdot\big((2+2\gamma)\mathbf v_i-(1+2\gamma)\mathbf v_j\big)\Big](\mathbf v_i-\mathbf v_j)
 + \frac{3+4\gamma}{2c^2}\sum_{j\neq i}\frac{\mu_j\,\mathbf a^N_j}{r_{ij}} ,
\end{aligned}
$$

with $\beta=\gamma=1$ in General Relativity. Three groups of terms, each with a meaning:

* **The bracket on the Newtonian attraction.** Gravity is weakened by the gravitational
  potential at the *field point* $i$ (the $\beta+\gamma$ sum: all bodies, not only the
  attractor) and at the *source* $j$ (the $2\beta-1$ sum: the Sun sits in Jupiter's potential),
  and modified by the velocities of both. Potentials and velocities enter because in GR
  energy — kinetic and potential — gravitates, and because the superposition of fields is
  non-linear ($\beta$). The last term carries the acceleration of the source: the field is not
  the instantaneous Newtonian one but is retarded, to this order.
* **The velocity-dependent force** along $\mathbf v_i-\mathbf v_j$ is the gravitomagnetic
  part: moving masses produce a field that acts on moving bodies like a magnetic field, with
  $\gamma$ fixing its strength. It is what turns the perihelion.
* **The source-acceleration term** $\sum\mu_j\mathbf a^N_j/r_{ij}$ is again retardation.

For a single mass at rest and a test particle every sum collapses to $\mu_0/r$, $v_0 = 0$,
$\mathbf a_0^N = 0$, and what is left is exactly Lesson 04's bracket,
$\frac{\mu_0}{c^2r^3}[(2(\beta+\gamma)\mu_0/r-\gamma v^2)\mathbf r+2(1+\gamma)(\mathbf r\cdot\mathbf v)\mathbf v]$
(`tests/test_eih.py` checks this identity for arbitrary $\beta,\gamma,c$).

**Why the ephemerides integrate EIH.** (i) *Consistency*: the equations are Lorentz-invariant
to $O(c^{-2})$, so they must be written in one inertial frame — the solar-system barycentre —
and the velocity terms of *all* bodies belong together; a Sun-only term with Galilean relative
velocities is a different (frame-dependent) force law. (ii) *The Sun moves*: in `make_pn1` the
Sun receives no relativistic acceleration at all; in EIH it does, and so do the planets from
one another. (iii) *Accuracy*: ranging to spacecraft measures distances to metres; the mutual
terms are far below Mercury's $43''$ but not below a metre. (iv) *Generality*: at 1PN the
relative orbit of *any* two masses advances by $6\pi G(m_1+m_2)/(c^2a(1-e^2))$ per orbit,
whatever the mass ratio (Robertson 1938) — the same equations serve binary pulsars. The
coordinate time of the ephemeris is TDB (barycentric), which runs at a different rate from
clocks on Earth; we note the fact and do not model it.

```python
import math
import time

import matplotlib.pyplot as plt
import numpy as np

from gravsim import constants as C
from gravsim.analysis import (
    fit_secular_rate,
    gr_precession_arcsec_per_century,
    perihelion_longitude_series,
    perihelion_rate_difference,
    separation,
    state_to_elements,
    wrap_angle,
)
from gravsim.bodies import SystemState, Trajectory
from gravsim.data.horizons import load_j2000, load_reference_series
from gravsim.forces import accel_newton, make_eih, make_newton_pn1, make_pn1
from gravsim.integrators import integrate_scipy
from gravsim.viz import PALETTE, figure_grid, plot_precession, plot_separation

plt.rcParams["figure.dpi"] = 100


def daily_grid(t0, years, step=1.0):
    return t0 + step * np.arange(int(round(years * C.JULIAN_YEAR_DAYS / step)) + 1)


def elements(state, body, central="Sun"):
    r, v = state.relative_to(central)
    i = state.index(body)
    gm = float(state.gm[state.index(central)] + state.gm[i])
    el = state_to_elements(r[i], v[i], gm)
    return {k: float(el[k]) for k in ("a", "e", "inc", "period")} | {"gm": gm}


def on_grid(traj, t_ref):
    """Sub-sample a daily run at the epochs of the (20-day) Horizons series."""
    idx = np.searchsorted(traj.t, t_ref)
    return Trajectory(traj.t[idx], traj.pos[idx], traj.vel[idx], traj.gm, traj.names)
```

## 2. Three runs on one grid

Sun + 8 planets from the J2000 snapshot, ten years, daily samples, DOP853 at the production
tolerances: Newton, the full EIH law, and Lesson 04's Newton + Sun-only 1PN term for comparison.
EIH is pure numpy over the $(N,N,3)$ separation tensor and costs about four Newtonian
(numba) runs; the adaptive integrator takes the same number of steps for all three, because
the relativistic terms are $10^{-8}$ of the force. The century versions of every number below
come from `scripts/validate.py` and the `slow` tests of `tests/test_eih.py`.

```python
state = load_j2000()
t_grid = daily_grid(state.t, 10.0)
laws = {
    "Newton": accel_newton,
    "EIH": make_eih(c=C.C_AU_DAY),
    "Newton + 1PN (Sun only)": make_newton_pn1(c=C.C_AU_DAY, central=state.index("Sun")),
}
runs = {}
for name, law in laws.items():
    tic = time.perf_counter()
    runs[name] = integrate_scipy(state, law, t_grid)
    print(f"{name:24s} {time.perf_counter() - tic:5.1f} s, {runs[name].meta['nfev']} force evaluations")
newton9, eih9, pn9 = runs["Newton"], runs["EIH"], runs["Newton + 1PN (Sun only)"]
```

## 3. The anomaly from data

Horizons (DE441) is our "observation": it was fitted to centuries of ranging and astrometry
and it contains GR. We sub-sample our runs at its 20-day epochs and difference the longitude
of perihelion, $\varpi_{\rm Horizons}(t) - \varpi_{\rm run}(t)$. Because the ephemeris and the
runs contain the same planets, the planetary wobble of tens of arcseconds (Lesson 02) cancels
in the difference, exactly as in the run-vs-run differences of Lesson 04 — only now one of the
"runs" is the real solar system.

The Newtonian residual is a straight line with the slope Le Verrier could not explain. The EIH
residual is flat: what remains ($0.02''$/cy) is the solar $J_2$, the asteroids, the Moon and
Pluto, which Horizons contains and our nine bodies do not. The right panel shows the same fact
as a distance: the Newtonian Mercury drifts away from the ephemeris by $\sim1500$ km in ten
years, the relativistic ones stay within a few km.

```python
ref = load_reference_series()
mask = ref.t <= t_grid[-1] + 1e-9
horizons = Trajectory(ref.t[mask], ref.pos[mask], ref.vel[mask], ref.gm, ref.names)
t_h, w_h = perihelion_longitude_series(horizons, "Mercury")

fig, axes = figure_grid(2, panel=(6.0, 4.0))
print(f"{'run':24s} {'Horizons − run [arcsec/cy]':>28s} {'final separation [km]':>22s}")
residual = {}
for name, traj in runs.items():
    sub = on_grid(traj, horizons.t)
    _, w = perihelion_longitude_series(sub, "Mercury")
    d = wrap_angle(w_h - w)
    fit = fit_secular_rate(t_h, d)
    sep = separation(sub, horizons, "Mercury", t_atol=1e-9)
    residual[name] = (fit.rate_arcsec_per_century, fit.rate_err_arcsec_per_century)
    print(f"{name:24s} {fit.rate_arcsec_per_century:15.3f} ± {fit.rate_err_arcsec_per_century:<10.3f} {sep[-1] * C.AU_KM:22.0f}")
    plot_precession(t_h, d, fit.rate if name == "Newton" else None, ax=axes[0], label=f"Horizons − {name}")
    plot_separation(horizons.t, sep, ax=axes[1], label=f"{name} vs Horizons")
axes[0].set_title("Mercury: ϖ(Horizons) − ϖ(run), Le Verrier's residual")
axes[1].set_yscale("log")
axes[1].set_title("distance of each run from the ephemeris")
print(f"\nreference: Einstein 1915 / Lesson 04        {C.MERCURY_PRECESSION_GR_ARCSEC_CY:7.2f}")
```

### Where the anomaly lives

A perihelion that lags by $\Delta\varpi$ is an ellipse rotated by $\Delta\varpi$; the planet on it
is displaced by $\sim r\,\Delta\varpi$ *along its track*, and the displacement grows linearly with
time. The 3-D view draws the vector from the Horizons Mercury to the Newtonian one, in km,
over the ten years (one dot per 20-day epoch of the ephemeris): a widening loop in the orbital
plane, with the EIH run's vector lost at the origin. The right panel checks the direction: the
displacement is almost entirely the component *opposite* to Mercury's velocity — the Newtonian
planet runs behind the real one, as a lagging perihelion requires.

```python
r_h, v_h = horizons.body("Mercury")
fig = plt.figure(figsize=(12.0, 5.0), layout="constrained")
ax3 = fig.add_subplot(1, 2, 1, projection="3d")
disp = {}
for k, name in enumerate(("Newton", "EIH")):
    d_km = (on_grid(runs[name], horizons.t).body("Mercury")[0] - r_h) * C.AU_KM
    disp[name] = d_km
    ax3.plot(d_km[:, 0], d_km[:, 1], d_km[:, 2], ".-", color=PALETTE[k], lw=0.6, ms=3, label=f"{name} − Horizons")
ax3.plot([0], [0], [0], "k*", ms=9, label="Horizons")
lim = np.abs(disp["Newton"]).max() * 1.05
ax3.set_xlim(-lim, lim); ax3.set_ylim(-lim, lim); ax3.set_zlim(-lim, lim); ax3.set_box_aspect((1, 1, 1))
ax3.set_xlabel("Δx [km]"); ax3.set_ylabel("Δy [km]"); ax3.set_zlabel("Δz [km]")
ax3.view_init(elev=35, azim=-50); ax3.legend(loc="upper left", fontsize=8)
ax3.set_title("Mercury: displacement of the runs from the ephemeris, 10 yr")

ax2 = fig.add_subplot(1, 2, 2)
t_yr = (horizons.t - horizons.t[0]) / C.JULIAN_YEAR_DAYS
d_n = disp["Newton"]
along = np.einsum("ij,ij->i", d_n, v_h) / np.linalg.norm(v_h, axis=1)  # component along the velocity
ax2.plot(t_yr, np.linalg.norm(d_n, axis=1), lw=1.2, color=PALETTE[0], label="|Newton − Horizons|")
ax2.plot(t_yr, -along, lw=0.8, color="k", ls="--", label="component opposite to the velocity")
ax2.set_xlabel("t [years]"); ax2.set_ylabel("displacement [km]")
ax2.grid(True, lw=0.3); ax2.legend(fontsize=8, loc="upper left")
ax2.set_title("the Newtonian Mercury falls behind along its track")
print(f"Newton − Horizons after 10 years: {np.linalg.norm(d_n[-1]):.0f} km, of which {-along[-1]:.0f} km behind along the track; "
      f"EIH − Horizons: {np.linalg.norm(disp['EIH'][-1]):.1f} km")
```

## 4. What the planets see: EIH minus Newton

The run-vs-run difference on the common grid gives the relativistic advance of every inner
planet, now with every body relativistic. Ten years is enough for Mercury and Mars; Venus and
the EMB are nearly circular, their perihelia are badly defined, and the periodic terms of the
difference do not average out in a decade (the century run of `tests/test_eih.py` gives
$8.637$, $3.851$ and $1.351$ against the analytic $8.62$, $3.84$, $1.35$).

```python
print(f"{'body':8s} {'EIH − Newton [arcsec/cy]':>26s} {'analytic 1PN':>13s}")
rates = {}
for name in ("Mercury", "Venus", "EMB", "Mars"):
    rate, err = perihelion_rate_difference(eih9, newton9, name)
    e = elements(state, name)
    rates[name] = (rate, err)
    print(f"{name:8s} {rate:16.3f} ± {err:6.3f} {float(gr_precession_arcsec_per_century(e['a'], e['e'], e['gm'])):13.3f}")
```

## 5. EIH versus the Sun-only term: why the difference is tiny, and what it changes

At J2000 the EIH correction to Mercury's acceleration and the Sun-only 1PN term differ by
$3\times10^{-4}$ of the correction itself — i.e. by $2\times10^{-11}$ of the force. The table
shows why. The extra ingredients of EIH are, in order of size: the Sun's barycentric velocity
($v_\odot/v_{\rm Mercury} = 4\times10^{-4}$, entering the velocity terms), the planets'
potentials at Mercury and at the Sun ($\sum\mu_p/r_p$ is $10^{-4}$ of $\mu_\odot/r$), the Sun's
Newtonian acceleration ($\sim\mu_{\rm Jup}/r_{\rm Jup}^2$), and the planets as *direct*
relativistic sources ($\mu_p/\mu_\odot\cdot(r/r_p)^2\sim10^{-6}$). Moreover the largest of these,
the velocity terms, are tied to the Sun's 12-year wobble around Jupiter and are **periodic**:
their secular effect on $\varpi$ averages to $\sim10^{-5}\,''$/cy, far below anything observable.

What EIH does change: the Sun itself is now accelerated relativistically (the Sun-only term
gives it nothing), the outer planets get corrections that differ by percent from the Sun-only
ones (the Sun sits in Jupiter's potential, which is $0.6\%$ of its own potential at Neptune),
and the Mercury positions differ by a few km per century — below the $43''$ story, above the
metre-level of ranging.

```python
a_n = accel_newton(state.pos, state.vel, state.gm, state.t)
a_eih = make_eih()(state.pos, state.vel, state.gm, state.t)
a_pn1 = make_pn1(central=state.index("Sun"))(state.pos, state.vel, state.gm, state.t)
print(f"{'body':8s} {'|a_EIH − a_N| / |a_N|':>22s} {'|a_1PN| / |a_N|':>16s} {'|a_EIH − a_N − a_1PN| / |a_1PN|':>32s}")
for i, name in enumerate(state.names):
    corr = a_eih[i] - a_n[i]
    n_n, n_p = np.linalg.norm(a_n[i]), np.linalg.norm(a_pn1[i])
    extra = f"{np.linalg.norm(corr - a_pn1[i]) / n_p:32.1e}" if n_p > 0 else f"{'(no Sun-only term at all)':>32s}"
    print(f"{name:8s} {np.linalg.norm(corr) / n_n:22.1e} {n_p / n_n:16.1e} {extra}")

v_sun = np.linalg.norm(state.vel[state.index("Sun")])
v_merc = np.linalg.norm(state.relative_to("Sun")[1][state.index("Mercury")])
phi_planets = sum(state.gm[j] / np.linalg.norm(state.pos[1] - state.pos[j]) for j in range(2, state.n))
phi_sun = C.GM_SUN / np.linalg.norm(state.relative_to("Sun")[0][1])
print(f"\nv_Sun / v_Mercury = {v_sun / v_merc:.1e};  planets' potential at Mercury / Sun's = {phi_planets / phi_sun:.1e}")
rate, err = perihelion_rate_difference(eih9, pn9, "Mercury")
sep_m = separation(eih9, pn9, "Mercury") * C.AU_KM
sep_s = separation(eih9, pn9, "Sun") * C.AU_KM
print(f"Mercury, EIH − (Newton + 1PN): secular rate {rate:+.1e} ± {err:.1e} arcsec/cy;  "
      f"separation after 10 years {sep_m[-1]:.2f} km (Sun: {sep_s[-1]:.3f} km)")

t, w_e = perihelion_longitude_series(eih9, "Mercury")
_, w_n = perihelion_longitude_series(newton9, "Mercury")
_, w_p = perihelion_longitude_series(pn9, "Mercury")
fig, axes = figure_grid(2, panel=(6.0, 4.0))
d = wrap_angle(w_e - w_n)
plot_precession(t, d, fit_secular_rate(t, d).rate, ax=axes[0], label="EIH − Newton")
axes[0].set_title("Mercury: the GR advance in the nine-body EIH run")
d2 = wrap_angle(w_e - w_p)
plot_precession(t, d2, fit_secular_rate(t, d2).rate, ax=axes[1], label="EIH − (Newton + 1PN, Sun only)")
axes[1].set_title("what the mutual terms add: periodic, no secular drift")
```

## 6. A binary star on Mercury's orbit

Where the two laws really part company is when the "planet" is heavy. At 1PN the relative orbit
of two masses advances by $6\pi G(m_1+m_2)/(c^2a(1-e^2))$ per orbit whatever the mass ratio —
the formula `gr_precession_arcsec_per_century` evaluates with $gm = \mu_\odot+\mu_{\rm planet}$.
We put a companion of growing mass on Mercury's relative orbit (same $a$ and $e$; the velocity
is rescaled by $\sqrt{M_{\rm new}/M_{\rm old}}$), reduce $c$ by 100 to keep the runs short, and
measure the advance with both laws. EIH stays on the formula; the Sun-only term, which treats
the companion as a test body in the field of *half* the mass and ignores the Sun's own motion,
does not. The small, mass-dependent deficits of the EIH ratios are second order in $\mu/p$
(Lesson 04 §6, Lesson 05 §4): at $c/100$ the orbit is no longer deep in the weak field.

```python
two = load_j2000().subset(("Sun", "Mercury")).to_barycentric()
r_rel, v_rel = two.pos[1] - two.pos[0], two.vel[1] - two.vel[0]
gm_old = float(two.gm.sum())


def binary(gm_companion, c, n_orbits=30):
    """Sun + a companion of mass gm_companion on Mercury's relative orbit: Newton, EIH, Sun-only 1PN."""
    gm = np.array([C.GM_SUN, gm_companion])
    v = v_rel * math.sqrt(gm.sum() / gm_old)  # same a and e for the new total mass
    pos = np.array([-gm[1] / gm.sum() * r_rel, gm[0] / gm.sum() * r_rel])
    vel = np.array([-gm[1] / gm.sum() * v, gm[0] / gm.sum() * v])
    st = SystemState(two.t, pos, vel, gm, ("Sun", "Companion"))
    el = elements(st, "Companion")
    t_eval = st.t + np.arange(0.0, n_orbits * el["period"], 0.25)
    laws = (("Newton", accel_newton), ("EIH", make_eih(c=c)), ("Sun-only 1PN", make_newton_pn1(c=c, central=0)))
    return st, el, {name: integrate_scipy(st, law, t_eval) for name, law in laws}


c_small = C.C_AU_DAY / 100.0
print(f"{'companion / Sun':>16s} {'EIH / analytic':>15s} {'Sun-only 1PN / analytic':>24s} {'v/c':>6s}")
for mass in (C.GM["Mercury"], 0.1 * C.GM_SUN, C.GM_SUN):
    st, el, out = binary(mass, c_small)
    analytic = float(gr_precession_arcsec_per_century(el["a"], el["e"], el["gm"], c=c_small))
    r_e, _ = perihelion_rate_difference(out["EIH"], out["Newton"], "Companion")
    r_p, _ = perihelion_rate_difference(out["Sun-only 1PN"], out["Newton"], "Companion")
    v_over_c = np.linalg.norm(st.relative_to("Sun")[1][1]) / c_small
    print(f"{mass / C.GM_SUN:16.4g} {r_e / analytic:15.4f} {r_p / analytic:24.4f} {v_over_c:6.3f}")
```

## 7. Summary

* Le Verrier's $38''$ (Newcomb's $43''$) per century was the residual between the best Newtonian
  model and the observations; Vulcan was the Neptune method applied once too often; Einstein's
  field equations gave the number with no free parameter.
* The EIH equations make every body a relativistic source: potentials of all bodies at the
  field point and at the source, velocity-dependent (gravitomagnetic) forces, source
  accelerations. Their test-particle limit is Lesson 04's term; the ephemerides integrate the
  full set for consistency, accuracy and generality.
* The anomaly from data: against Horizons, the Newtonian Mercury's perihelion lags at
  $43.0''$/cy and the planet falls behind along its track by $\sim1500$ km per decade; the EIH
  run's residual is $0.02''$/cy and a few km.
* For Mercury's secular rate EIH and the Sun-only term agree to $10^{-5}\,''$/cy, because the
  extra terms are $3\times10^{-4}$ of the correction and periodic. For comparable masses only
  EIH gives the mass-ratio-independent $6\pi G(m_1+m_2)/(c^2p)$.
