# Lesson 09 — What else competes with GR?

**Phase 8 of the gravsim project: the precession budget at the 0.01″ level.**

Lessons 04–07 extracted the $42.98''$ per century of General Relativity from a Newtonian
background of $531.6''$. Ranging to Mercury — radar since the 1960s, the MESSENGER spacecraft
from 2011 to 2015 — now measures the advance to about $0.01''$ per century, and at that level
other things move the perihelion: the quadrupole of the rotating Sun, the dragging of inertial
frames by its spin, the asteroid belt, and the way we model the Earth–Moon pair. This lesson
puts a number on each with the force laws of `gravsim.forces.extras` and the Earth + Moon
ephemeris of Phase 8, and asks what, besides the perihelion, tells them apart.

By the end you should be able to

1. write down Mercury's precession budget and say which entries are Newtonian, which are
   relativistic, and which are modelling choices;
2. derive the secular node and perihelion rates of an orbit about an oblate body (Kaula) and
   about a spinning one (Lense–Thirring), and verify them on amplified runs;
3. explain why Mercury's perihelion alone cannot separate the solar $J_2$ from the PPN
   combination $2\gamma-\beta$, and what does;
4. measure the largest precession in the solar system — the lunar perigee, $40.7°$ per year —
   and explain why it is purely Newtonian; and say by how much the Earth–Moon barycentre
   fails to move like a point mass.

## 1. The physics: the budget and the three competitors

### 1.1 The budget

In an inertial frame Mercury's perihelion advances by $574.1\pm0.4''$ per century (Lesson 06).
Where it comes from, in decreasing order:

| contribution | ″/century | nature | section |
|---|---|---|---|
| the planets | $531.6$ | Newtonian | Lesson 02 |
| General Relativity | $42.98$ | $6\pi GM/(c^2a(1-e^2))$ per orbit | Lessons 04–07 |
| solar quadrupole, $J_2 = 2.2\times10^{-7}$ | $+0.028$ | Newtonian — the Sun's shape | §4 |
| Earth and Moon as one body at their barycentre | $-0.0025$ | a modelling choice | §3 |
| Lense–Thirring frame dragging | $-0.002$ | relativistic — the Sun's spin | §5 |
| the asteroid belt ($1.2\times10^{-9}\,M_\odot$ at $2.7$ AU) | $\sim+0.0014$ | Newtonian, not modelled | §2 |

The bound $|2\gamma-\beta-1| < 8\times10^{-5}$ of Lesson 07 is a statement about the
advance to $3\times8\times10^{-5}\times42.98'' = 0.010''$/cy. The first two lines are not enough
at that level: the solar $J_2$ is three times the precision, and it is the first competitor.

### 1.2 An oblate Sun

A body flattened by its rotation has, to the second zonal harmonic, the potential per unit
mass (Kaula 1966; Murray & Dermott 1999, §6.11)

$$
\Phi = -\frac{GM}{r}\Big[1 - J_2\Big(\frac{R}{r}\Big)^2 P_2(\cos\theta)\Big],
\qquad P_2(x) = \tfrac12(3x^2-1),
$$

with $\theta$ the colatitude from the spin axis $\hat{\mathbf n}$. Its gradient, with
$s = \mathbf r\cdot\hat{\mathbf n}$ and $k = GMJ_2R^2$, is the acceleration of `make_j2`:

$$
\mathbf a_{J_2} = \frac{3k}{2r^5}\Big[\Big(5\frac{s^2}{r^2}-1\Big)\mathbf r - 2s\,\hat{\mathbf n}\Big],
$$

stronger than Newton's pull over the equator, weaker over the poles. Averaged over an orbit
with Gauss's planetary equations (`docs/theory/extras.md`) it gives the classical secular
rates — with $n$ the mean motion, $p = a(1-e^2)$ and $i$ the inclination **to the Sun's
equator** (Mercury: $3.38°$, not the $7.0°$ to the ecliptic; the solar pole is tilted $7.25°$,
`constants.SUN_POLE_ECLIPTIC`):

$$
\dot\Omega = -\tfrac32\,nJ_2\Big(\tfrac Rp\Big)^2\cos i, \qquad
\dot\omega = \tfrac34\,nJ_2\Big(\tfrac Rp\Big)^2(5\cos^2 i-1) .
$$

For Mercury $\dot\Omega = -0.028''$, $\dot\omega = +0.056''$ and
$\dot\varpi = \dot\Omega+\dot\omega = +0.028''$ per century; $a$, $e$ and $i$ have no secular
change. Lesson 03 used this term with $J_2 = 10^{-3}$ and the ecliptic pole as the axis, to
calibrate the instrument; here the axis is the real one and the size is the real one.

**The degeneracy.** Both GR and $J_2$ advance $\varpi$. From Mercury's perihelion alone a measured
advance $\Delta = \tfrac{2+2\gamma-\beta}{3}\times42.98'' + 0.028''\times J_2/(2.2\times10^{-7})$
cannot be split: $J_2$ and $2\gamma-\beta$ are degenerate (Lesson 07 §5). Three things break
it. (i) *The node*: GR advances $\omega$ only, while $J_2$ regresses $\Omega$ by as much as it
advances $\varpi$. (ii) *Other planets*: the GR rate scales as
$n/(c^2p) \propto a^{-5/2}(1-e^2)^{-1}$, the $J_2$ rate as $nJ_2(R/p)^2 \propto a^{-7/2}(1-e^2)^{-2}$,
so their ratio falls as $1/p$: $6.5\times10^{-4}$ at Mercury, $3.3\times10^{-4}$ at Venus,
$1.6\times10^{-4}$ at Mars — two planets give two equations. (iii) *Helioseismology* measures
$J_2 = (2.2\pm0.1)\times10^{-7}$ from the Sun's internal rotation, and the MESSENGER ranging
solution (Park et al. 2017) fitted $J_2 = 2.25\times10^{-7}$ together with $\beta$.

### 1.3 A spinning Sun: frame dragging

To 1PN order a rotating mass drags inertial frames. Its angular momentum $\mathbf J$ produces a
*gravitomagnetic* field that acts on a moving body like a magnetic dipole on a charge, with the
sign reversed because like masses attract (Lense & Thirring 1918; Soffel 1989):

$$
\mathbf a_{LT} = \mathbf v\times\mathbf B, \qquad
\mathbf B = \frac{2}{c^2r^3}\big[G\mathbf J - 3(\hat{\mathbf r}\cdot G\mathbf J)\,\hat{\mathbf r}\big]
$$

(`make_lense_thirring`; $GJ_\odot = 1.1\times10^{-10}$ AU⁵/day³ for the helioseismic
$J_\odot = 1.9\times10^{41}$ kg m² s⁻¹). The force does no work ($\mathbf a\cdot\mathbf v = 0$)
and its secular effect (`docs/theory/extras.md`) is

$$
\dot\Omega = \frac{2GJ}{c^2a^3(1-e^2)^{3/2}}, \qquad
\dot\omega = -\frac{6GJ\cos i}{c^2a^3(1-e^2)^{3/2}} :
$$

the node is dragged *in the sense of the spin* whatever the inclination, and the argument of
perihelion is driven *back* three times faster. For Mercury: $+0.001''$, $-0.003''$,
$\dot\varpi = -0.002''$ per century — a hundred times below the ranging precision. Around the
Earth it has been measured: the nodes of the LAGEOS satellites are dragged by $31$ mas/yr
(Ciufolini & Pavlis 2004, to 10 %; with LARES to a few %), and the gyroscopes of Gravity
Probe B precessed by $37\pm7$ mas/yr against the predicted $39$ (Everitt et al. 2011).

### 1.4 The Moon

The default system of this project replaces Earth and Moon by one body at their barycentre
(the EMB). Phase 8 ships the J2000 state and a ten-year daily Horizons series with Earth and
Moon separate (`load_j2000(with_moon=True)`, `load_moon_reference_series`). Two things to check.

*Does the EMB approximation hold?* Not quite. The Sun's pull on a dumbbell is not the pull on
its barycentre: expanding to second order in the separation $\boldsymbol\rho$, the barycentre
feels $\tfrac12 w_Ew_M(\boldsymbol\rho\cdot\nabla)^2\mathbf a$ in excess, with
$w_Ew_M = m_Em_M/(m_E+m_M)^2 = 0.0120$. Averaged over a month this is an extra *inward* pull

$$
\varepsilon = \tfrac34\,w_Ew_M\Big(\frac{a_{\rm Moon}}{a_{\rm Earth}}\Big)^2 = 5.9\times10^{-8}
$$

of the solar one — as if the Sun were heavier by that fraction — which, for the same initial
state, raises the mean motion by $2\varepsilon$ and makes the real barycentre run ahead of the
point mass by $2\varepsilon\,n\,a\,t \approx 110$ km per year. Negligible for Mercury's
perihelion ($-0.0025''$/cy, measured below); not negligible for the Earth's position.

*The largest precession in the solar system.* The Moon's perigee advances by $40.69°$ per year
and its node regresses by $19.34°$ per year — $25\,000$ times Mercury's whole advance — and all
of it is Newtonian: the Sun's tide on the lunar orbit. To first order in
$m = n_\odot/n_{\rm Moon} \approx 0.075$ both rates are $\tfrac34 m^2 n = 20.1°$/yr (Newton 1687,
who got the node right and only half the motion of the perigee). The higher orders double the
perigee rate (Clairaut 1749 — the episode that nearly had the inverse-square law amended); the
series (Delaunay 1860; Brouwer & Clemence 1961, ch. 12)

$$
\dot\varpi = n\Big[\tfrac34m^2 + \tfrac{225}{32}m^3 + \tfrac{4071}{128}m^4 + \tfrac{265493}{2048}m^5 + \dots\Big],
\qquad
\dot\Omega = n\Big[-\tfrac34m^2 + \tfrac{9}{32}m^3 + \tfrac{273}{128}m^4 + \tfrac{9797}{2048}m^5 + \dots\Big]
$$

converges slowly toward the observed values. The simulation needs none of this: it integrates
Sun, planets, Earth and Moon and reads the rates off the osculating geocentric elements.

```python
import math
import time

import matplotlib.pyplot as plt
import numpy as np

from gravsim import constants as C
from gravsim.analysis import (
    element_series,
    fit_secular_rate,
    gr_precession_arcsec_per_century,
    inclination_to_axis,
    perihelion_rate_difference,
    rate_arcsec_per_century,
    separation,
    state_to_elements,
    wrap_angle,
)
from gravsim.bodies import Trajectory
from gravsim.data.horizons import load_j2000, load_moon_reference_series
from gravsim.forces import accel_newton, combine, make_j2, make_lense_thirring, make_newton_pn1
from gravsim.integrators import integrate_scipy
from gravsim.viz import PALETTE, figure_grid

plt.rcParams["figure.dpi"] = 100
DEG_YR = C.DEG_PER_RAD * C.JULIAN_YEAR_DAYS  # rad/day -> degrees per Julian year


def daily_grid(t0, years):
    return t0 + np.arange(int(round(years * C.JULIAN_YEAR_DAYS)) + 1, dtype=float)


def equator_frame(pole=C.SUN_POLE_ECLIPTIC):
    """Rows x, y, z of the solar-equator frame: z = the pole, x = the equator's node on the ecliptic."""
    z = np.asarray(pole) / np.linalg.norm(pole)
    x = np.cross([0.0, 0.0, 1.0], z)
    x /= np.linalg.norm(x)
    return np.stack([x, np.cross(z, x), z])


def rotate(traj, R):
    return Trajectory(traj.t, traj.pos @ R.T, traj.vel @ R.T, traj.gm, traj.names, dict(traj.meta))


def differenced_rates(trajA, trajB, body="Mercury", keys=("raan", "argp", "varpi"), R=None):
    """{key: (rate, err, series)} of element_A - element_B (wrapped, fitted), optionally in a rotated frame."""
    if R is not None:
        trajA, trajB = rotate(trajA, R), rotate(trajB, R)
    out = {}
    for key in keys:
        t, xa = element_series(trajA, body, key)
        _, xb = element_series(trajB, body, key)
        d = wrap_angle(xa - xb)
        fit = fit_secular_rate(t, d)
        out[key] = (fit.rate_arcsec_per_century, fit.rate_err_arcsec_per_century, d)
    return t, out


def elements(state, body, central="Sun"):
    r, v = state.relative_to(central)
    i = state.index(body)
    gm = float(state.gm[state.index(central)] + state.gm[i])
    el = state_to_elements(r[i], v[i], gm)
    out = {k: float(el[k]) for k in ("a", "e", "inc", "period")} | {"gm": gm}
    out["inc_eq"] = float(inclination_to_axis(r[i], v[i], C.SUN_POLE_ECLIPTIC))  # to the solar equator
    return out


def kaula(el, j2=C.SUN_J2, r_eq=C.R_SUN_AU):
    """Secular (node, argument) rates [rad/day] about an oblate body; i relative to its equator."""
    n = math.sqrt(el["gm"] / el["a"] ** 3)
    k = n * j2 * (r_eq / (el["a"] * (1.0 - el["e"] ** 2))) ** 2
    ci = math.cos(el["inc_eq"])
    return -1.5 * k * ci, 0.75 * k * (5.0 * ci * ci - 1.0)


def lense_thirring(el, gj=C.GJ_SUN, c=C.C_AU_DAY):
    """Secular (node, argument) rates [rad/day] from frame dragging by the spin GJ."""
    k = gj / (c * c * el["a"] ** 3 * (1.0 - el["e"] ** 2) ** 1.5)
    return 2.0 * k, -6.0 * k * math.cos(el["inc_eq"])


R_EQ = equator_frame()
```

## 2. The budget in one picture

Six contributions spanning six orders of magnitude, on a log axis, with the precision of
today's ranging marked. The two solar terms and the ring are the analytic rates of §1 for
Mercury's osculating elements at J2000; the Moon row is measured in §3.

```python
state = load_j2000()
el_m = elements(state, "Mercury")
j2_raan, j2_argp = kaula(el_m)
lt_raan, lt_argp = lense_thirring(el_m)
n_merc = math.sqrt(el_m["gm"] / el_m["a"] ** 3)
ring = rate_arcsec_per_century(0.75 * n_merc * 1.2e-9 * (el_m["a"] / 2.7) ** 3)  # (3/4) n (m/M) (a/a_ring)^3
budget = {
    "planets (Newtonian, Lesson 02)": (C.MERCURY_PRECESSION_NEWTON_ARCSEC_CY, "Clemence 1947"),
    "General Relativity (Lessons 04-07)": (C.MERCURY_PRECESSION_GR_ARCSEC_CY, "6πGM/(c²a(1−e²))"),
    "solar J2 = 2.2e-7": (float(rate_arcsec_per_century(j2_raan + j2_argp)), "Kaula, §4"),
    "Moon as a body vs the EMB": (-0.0025, "measured in §3"),
    "Lense-Thirring": (float(rate_arcsec_per_century(lt_raan + lt_argp)), "frame dragging, §5"),
    "asteroid belt (not modelled)": (float(ring), "ring of 1.2e-9 M☉ at 2.7 AU"),
}
print(f"{'contribution to Mercury':36s} {'arcsec/cy':>10s}   source")
for name, (rate, src) in budget.items():
    print(f"{name:36s} {rate:+10.4f}   {src}")
modelled = sum(v[0] for k, v in budget.items() if "not modelled" not in k)
print(f"{'sum of the modelled terms':36s} {modelled:+10.4f}   observed {C.MERCURY_PRECESSION_OBSERVED_ARCSEC_CY} ± 0.4")

fig, axes = figure_grid(1, panel=(8.5, 3.8))
ax = axes[0]
names = list(budget)[::-1]
rates = np.array([budget[k][0] for k in names])
colors = [PALETTE[0] if r > 0 else PALETTE[3] for r in rates]
colors[0] = "0.65"  # the unmodelled ring
ax.barh(names, np.abs(rates), color=colors, height=0.6)
for y, r in enumerate(rates):
    ax.text(abs(r) * 1.35, y, f"{r:+.4g}″/cy", va="center", fontsize=8)
ax.axvline(0.010, color="k", ls=":", lw=0.8)
ax.text(0.0115, 5.35, "ranging precision ~0.01″/cy", fontsize=7)
ax.set_xscale("log")
ax.set_xlim(3e-4, 3e4)
ax.set_xlabel("|secular rate of ϖ| [arcsec per century]")
ax.set_title("Mercury's perihelion budget (blue advances, red regresses, grey not modelled)")
```

## 3. Experiment 1: the Moon as a body

Four ten-year runs on the daily grid of the Horizons Moon series: the split system (Sun,
planets, Earth, Moon — ten bodies) and the default EMB system, each with Newton and with
Newton + the Sun's 1PN term. The Moon's 27-day orbit forces the adaptive integrator to steps of
half a day, so the split runs cost twice the force evaluations of Lesson 06's nine-body runs.
Then three questions: does Mercury care, does the barycentre of the pair move like the EMB,
and what does the lunar orbit itself do.

```python
ref = load_moon_reference_series()
t_grid = ref.t[: int(round(10.0 * C.JULIAN_YEAR_DAYS)) + 1]  # the daily Horizons grid, ten years
horizons = Trajectory(t_grid, ref.pos[: t_grid.size], ref.vel[: t_grid.size], ref.gm, ref.names)
split = load_j2000(with_moon=True)  # Sun, Mercury, Venus, Earth, Moon, Mars, Jupiter, Saturn, Uranus, Neptune
emb = load_j2000()  # Sun, Mercury, Venus, EMB, Mars, Jupiter, Saturn, Uranus, Neptune
runs = {}
for name, st, law in (
    ("split Newton", split, accel_newton),
    ("split GR", split, make_newton_pn1(central=0)),
    ("EMB Newton", emb, accel_newton),
    ("EMB GR", emb, make_newton_pn1(central=0)),
):
    tic = time.perf_counter()
    runs[name] = integrate_scipy(st, law, t_grid)
    print(f"{name:13s} {time.perf_counter() - tic:4.1f} s, {runs[name].meta['nfev']:6d} force evaluations")

# 1. Mercury: the GR difference is unchanged, the Newtonian rate shifts by a little
gr_split = perihelion_rate_difference(runs["split GR"], runs["split Newton"], "Mercury")
gr_emb = perihelion_rate_difference(runs["EMB GR"], runs["EMB Newton"], "Mercury")
shift = perihelion_rate_difference(runs["split Newton"], runs["EMB Newton"], "Mercury")
print(f"\nMercury, GR − Newton:  Earth + Moon {gr_split[0]:.4f} ± {gr_split[1]:.4f},  EMB {gr_emb[0]:.4f} ± {gr_emb[1]:.4f} arcsec/cy"
      f"  (difference {gr_split[0] - gr_emb[0]:.1e})")
print(f"Mercury, Newton(Earth + Moon) − Newton(EMB): {shift[0]:+.4f} ± {shift[1]:.4f} arcsec/cy   ← the 'Moon vs EMB' row of the budget")

# 2. the EMB approximation: the barycentre of the pair versus the point mass
iE, iM = split.index("Earth"), split.index("Moon")
w = split.gm[[iE, iM]] / split.gm[[iE, iM]].sum()


def barycentre(traj):
    return w[0] * traj.body("Earth")[0] + w[1] * traj.body("Moon")[0]


drift = np.linalg.norm(barycentre(runs["split GR"]) - runs["EMB GR"].body("EMB")[0], axis=1) * C.AU_KM
vs_horizons = np.linalg.norm(barycentre(runs["split GR"]) - barycentre(horizons), axis=1) * C.AU_KM
emb_vs_horizons = np.linalg.norm(runs["EMB GR"].body("EMB")[0] - barycentre(horizons), axis=1) * C.AU_KM
el_e = elements(split, "Earth")
el_moon = elements(split, "Moon", central="Earth")
eps = 0.75 * w[0] * w[1] * (el_moon["a"] / el_e["a"]) ** 2
n_e = math.sqrt(el_e["gm"] / el_e["a"] ** 3)
predicted_drift = 2.0 * eps * n_e * (t_grid[-1] - t_grid[0]) * el_e["a"] * C.AU_KM
print(f"\nEarth–Moon barycentre after 10 yr: split run vs EMB run {drift[-1]:.0f} km"
      f"  (2εnat with ε = (3/4) w_E w_M (a_M/a_E)² = {eps:.2e}: {predicted_drift:.0f} km)")
print(f"                                   split run vs Horizons max {vs_horizons.max():.1f} km,"
      f"  EMB run vs Horizons max {emb_vs_horizons.max():.0f} km")
```

```python
# 3. the lunar orbit: osculating geocentric elements, gm = GM_Earth + GM_Moon
def lunar_fits(traj):
    r, v = traj.relative("Moon", "Earth")
    el = state_to_elements(r, v, float(traj.gm[traj.index("Earth")] + traj.gm[traj.index("Moon")]))
    n_sid = fit_secular_rate(traj.t, el["varpi"] + el["M"]).rate  # mean motion of the mean longitude
    fw, fo = fit_secular_rate(traj.t, el["varpi"]), fit_secular_rate(traj.t, el["raan"])
    return {"month": 2.0 * math.pi / n_sid, "perigee": fw.rate * DEG_YR, "perigee_err": fw.rate_err * DEG_YR,
            "node": fo.rate * DEG_YR, "node_err": fo.rate_err * DEG_YR, "varpi": el["varpi"], "raan": el["raan"],
            "a": float(el["a"].mean()), "e": float(el["e"].mean()), "inc": float(el["inc"].mean())}


moon_run, moon_h = lunar_fits(runs["split Newton"]), lunar_fits(horizons)
n_moon = math.sqrt(el_moon["gm"] / moon_run["a"] ** 3)
m = n_e / n_moon
lead = 0.75 * m**2 * n_moon * DEG_YR
peri_series = n_moon * DEG_YR * (0.75 * m**2 + 225 / 32 * m**3 + 4071 / 128 * m**4 + 265493 / 2048 * m**5)
node_series = n_moon * DEG_YR * (-0.75 * m**2 + 9 / 32 * m**3 + 273 / 128 * m**4 + 9797 / 2048 * m**5)
print(f"{'':24s} {'this run (10 yr)':>20s} {'Horizons, same span':>20s} {'secular reference':>18s}")
print(f"{'sidereal month [d]':24s} {moon_run['month']:20.4f} {moon_h['month']:20.4f} {27.3217:18.4f}")
print(f"{'perigee rate [deg/yr]':24s} {moon_run['perigee']:+13.3f} ± {moon_run['perigee_err']:.3f} {moon_h['perigee']:+20.3f} {C.MOON_PERIGEE_RATE_DEG_YR:+18.2f}")
print(f"{'node rate [deg/yr]':24s} {moon_run['node']:+13.3f} ± {moon_run['node_err']:.3f} {moon_h['node']:+20.3f} {C.MOON_NODE_RATE_DEG_YR:+18.2f}")
print(f"\nm = n☉/n☾ = {m:.4f}: first order (3/4) m² n = ±{lead:.2f} deg/yr;  series to m⁵: perigee {peri_series:+.2f}, node {node_series:+.2f} deg/yr")
print(f"mean distance {moon_run['a'] * C.AU_KM:.0f} km, e = {moon_run['e']:.4f}, inclination to the ecliptic {math.degrees(moon_run['inc']):.2f}°")
mercury_deg_yr = C.MERCURY_PRECESSION_OBSERVED_ARCSEC_CY / 3600.0 / 100.0
print(f"for comparison, Mercury's whole advance is {mercury_deg_yr:.4f} deg/yr: the lunar perigee moves {C.MOON_PERIGEE_RATE_DEG_YR / mercury_deg_yr:.0f} times faster")

sep_moon = {name: separation(runs[name], horizons, "Moon") * C.AU_KM for name in ("split Newton", "split GR")}
geo = np.linalg.norm(runs["split GR"].relative("Moon", "Earth")[0] - horizons.relative("Moon", "Earth")[0], axis=1) * C.AU_KM
print(f"Moon vs Horizons, barycentric: Newton max {sep_moon['split Newton'].max():.0f} km, Newton + 1PN max {sep_moon['split GR'].max():.0f} km;"
      f"  geocentric (GR run) max {geo.max():.0f} km")

t_yr = (t_grid - t_grid[0]) / C.JULIAN_YEAR_DAYS
fig, axes = figure_grid(2, panel=(6.0, 4.0))
ax = axes[0]
for key, label, k in (("varpi", "perigee ϖ", 0), ("raan", "node Ω", 1)):
    ax.plot(t_yr, np.degrees(np.unwrap(moon_run[key]) - moon_run[key][0]), color=PALETTE[k], lw=1.0, label=f"{label}, run")
    ax.plot(t_yr, np.degrees(np.unwrap(moon_h[key]) - moon_h[key][0]), color="k", lw=0.6, ls="--", label=f"{label}, Horizons")
ax.set_xlabel("t [years]"); ax.set_ylabel("angle − initial value [deg]"); ax.grid(True, lw=0.3); ax.legend(fontsize=8)
ax.set_title("the lunar perigee advances 40.7°/yr, the node regresses 19.3°/yr")
ax = axes[1]
ax.plot(t_yr, sep_moon["split Newton"], lw=0.8, color=PALETTE[0], label="Moon: Newton vs Horizons")
ax.plot(t_yr, sep_moon["split GR"], lw=0.8, color=PALETTE[1], label="Moon: Newton + 1PN vs Horizons")
ax.plot(t_yr, drift, lw=1.2, color=PALETTE[2], label="Earth–Moon barycentre: split run vs EMB run")
ax.plot(t_yr, vs_horizons, lw=0.8, color="k", ls="--", label="barycentre: split run vs Horizons")
ax.set_xlabel("t [years]"); ax.set_ylabel("separation [km]"); ax.set_yscale("log"); ax.grid(True, lw=0.3); ax.legend(fontsize=7)
ax.set_title("the EMB approximation drifts along the track at ~110 km/yr")
```

### Reading the numbers

* Mercury's GR difference is the same to $10^{-7}\,''$/cy whether the Earth is one body or two;
  its *Newtonian* rate shifts by $-0.0025''$/cy, the size of the Lense–Thirring term and well
  below the ranging precision. The EMB approximation is fine for Mercury.
* It is not fine for the Earth: the barycentre of the pair runs ahead of the point mass by
  $\sim1200$ km in ten years, within a few percent of $2\varepsilon nat$. The split run tracks
  Horizons' barycentre to a few km while the EMB run drifts away from it — this is why the
  ephemerides integrate Earth and Moon separately.
* The lunar orbit: the sidereal month to $10^{-5}$ d, the perigee and node rates to
  $0.01°$/yr of what Horizons gives *over the same span* (both fits carry the same long-period
  bias; the secular values need decades). The Moon stays within $160$ km of the ephemeris in
  ten years; what is missing is mostly the Earth's oblateness and the tides, not GR.

## 4. Experiment 2: solar oblateness

Sun + Mercury, ten years, daily samples, three runs on one grid besides Newton: $J_2$
exaggerated a thousandfold (the rates become $\sim30''$/cy, comfortably measurable), the real
$J_2 = 2.2\times10^{-7}$, and Newton + 1PN for comparison. The node and the argument of
perihelion are taken in the **solar-equator frame** (a rotation of the trajectory so that $z$
is the solar pole) and differenced against the Newtonian run, which cancels everything but the
$J_2$ term. The run reproduces Kaula's rates to $10^{-4}$ even at the real size: differencing
on a common grid removes the integrator's error too.

```python
two = load_j2000().subset(("Sun", "Mercury")).to_barycentric()
t2 = daily_grid(two.t, 10.0)
newton2 = integrate_scipy(two, accel_newton, t2)
j2_big = integrate_scipy(two, combine(accel_newton, make_j2(j2=1000 * C.SUN_J2)), t2)
j2_real = integrate_scipy(two, combine(accel_newton, make_j2()), t2)
gr2 = integrate_scipy(two, make_newton_pn1(central=0), t2)
el2 = elements(two, "Mercury")
print(f"Mercury: a = {el2['a']:.4f} AU, e = {el2['e']:.4f}, inclination {math.degrees(el2['inc']):.2f}° to the ecliptic,"
      f" {math.degrees(el2['inc_eq']):.2f}° to the solar equator")
print(f"\n{'run':20s} {'element':8s} {'fit [arcsec/cy]':>24s} {'Kaula':>10s} {'ratio':>7s}")
j2_results = {}
for name, traj, j2 in (("J2 × 1000", j2_big, 1000 * C.SUN_J2), ("J2 = 2.2e-7 (real)", j2_real, C.SUN_J2)):
    t, fits = differenced_rates(traj, newton2, R=R_EQ)
    pred = dict(zip(("raan", "argp"), kaula(el2, j2=j2)))
    pred["varpi"] = pred["raan"] + pred["argp"]
    j2_results[name] = (t, fits, pred)
    for key in ("raan", "argp", "varpi"):
        p = float(rate_arcsec_per_century(pred[key]))
        print(f"{name:20s} {key:8s} {fits[key][0]:13.5f} ± {fits[key][1]:<9.5f} {p:10.5f} {fits[key][0] / p:7.4f}")
_, fits_gr = differenced_rates(gr2, newton2, R=R_EQ)
_, fits_ecl = differenced_rates(j2_big, newton2, keys=("raan",))
print(f"\n(Newton + 1PN) − Newton:  node {fits_gr['raan'][0]:+.5f},  argument {fits_gr['argp'][0]:+.3f} arcsec/cy   → GR leaves the node alone")
print(f"J2 × 1000 moves the node on the ecliptic too: {fits_ecl['raan'][0]:+.3f} arcsec/cy  (real J2: {fits_ecl['raan'][0] / 1000:+.5f})")

t, fits, pred = j2_results["J2 × 1000"]
t_yr = (t - t[0]) / C.JULIAN_YEAR_DAYS
fig, axes = figure_grid(2, panel=(6.0, 4.0))
for ax, key, label in ((axes[0], "raan", "node Ω"), (axes[1], "argp", "argument of perihelion ω")):
    d = np.unwrap(fits[key][2]) * C.ARCSEC_PER_RAD
    ax.plot(t_yr, d - d[0], lw=0.8, color=PALETTE[0], label="(Newton + J2 × 1000) − Newton")
    p = float(rate_arcsec_per_century(pred[key]))
    ax.plot(t_yr, p * t_yr / 100.0, "k--", lw=1.0, label=f"Kaula: {p:+.2f}″/cy")
    ax.set_xlabel("t [years]"); ax.set_ylabel(f"Δ{label.split()[-1]} [arcsec]"); ax.grid(True, lw=0.3); ax.legend(fontsize=8)
    ax.set_title(f"{label}, solar-equator frame")

print(f"\n{'planet':8s} {'GR':>8s} {'J2':>9s} {'J2/GR':>9s} {'1/p [1/AU]':>11s}    secular rates of ϖ [arcsec/cy]; the ratio falls as 1/(a(1−e²))")
for body in ("Mercury", "Venus", "EMB", "Mars"):
    e = elements(state, body)
    gr_rate = float(gr_precession_arcsec_per_century(e["a"], e["e"], e["gm"]))
    j2_rate = float(rate_arcsec_per_century(sum(kaula(e))))
    print(f"{body:8s} {gr_rate:8.3f} {j2_rate:9.5f} {j2_rate / gr_rate:9.2e} {1.0 / (e['a'] * (1.0 - e['e'] ** 2)):11.3f}")
```

## 5. Experiment 3: frame dragging

The same protocol with the Lense–Thirring force: the solar spin exaggerated $10^5$ times (the
node is then dragged by $100''$/cy, the argument driven back by $300''$/cy) and the real
$J_\odot$. The run follows the formulas to $10^{-4}$; the real effect, $-0.002''$/cy on $\varpi$,
is resolved by the differencing even though the force is $\sim10^{-12}$ of Newton's —
$5\times10^4$ times smaller than the 1PN term. The same formula applied to the Earth's spin and the
LAGEOS orbit gives the $31$ mas/yr that laser ranging measured.

```python
lt_big = integrate_scipy(two, combine(accel_newton, make_lense_thirring(scale=1e5)), t2)
lt_real = integrate_scipy(two, combine(accel_newton, make_lense_thirring()), t2)
print(f"{'run':16s} {'element':8s} {'fit [arcsec/cy]':>24s} {'Lense–Thirring':>15s} {'ratio':>7s}")
lt_results = {}
for name, traj, gj in (("J × 1e5", lt_big, 1e5 * C.GJ_SUN), ("J real", lt_real, C.GJ_SUN)):
    t, fits = differenced_rates(traj, newton2, R=R_EQ)
    pred = dict(zip(("raan", "argp"), lense_thirring(el2, gj=gj)))
    pred["varpi"] = pred["raan"] + pred["argp"]
    lt_results[name] = (t, fits, pred)
    for key in ("raan", "argp", "varpi"):
        p = float(rate_arcsec_per_century(pred[key]))
        print(f"{name:16s} {key:8s} {fits[key][0]:13.5f} ± {fits[key][1]:<9.5f} {p:15.5f} {fits[key][0] / p:7.4f}")

a_n = accel_newton(two.pos, two.vel, two.gm, 0.0)
a_gr = make_newton_pn1(central=0)(two.pos, two.vel, two.gm, 0.0) - a_n
a_j2 = make_j2()(two.pos, two.vel, two.gm, 0.0)
a_lt = make_lense_thirring()(two.pos, two.vel, two.gm, 0.0)
norm = np.linalg.norm
print(f"\naccelerations on Mercury at J2000 relative to Newton's: 1PN {norm(a_gr[1]) / norm(a_n[1]):.1e},"
      f" J2 {norm(a_j2[1]) / norm(a_n[1]):.1e}, Lense–Thirring {norm(a_lt[1]) / norm(a_n[1]):.1e}")

# the Earth's spin on LAGEOS: J_earth = 5.86e33 kg m^2/s, a = 12 270 km, e = 0.0045
gj_earth = C.G_SI * 5.86e33 * C.DAY_S**3 / (C.AU_KM * 1e3) ** 5
a_lageos = 12_270.0 / C.AU_KM
node_lageos = 2.0 * gj_earth / (C.C_AU_DAY**2 * a_lageos**3 * (1.0 - 0.0045**2) ** 1.5)  # rad/day
print(f"LAGEOS node, Lense–Thirring: {node_lageos * DEG_YR * 3600e3:.1f} mas/yr  (measured 1998-2004: ~31 mas/yr to 10 %)")

t, fits, pred = lt_results["J × 1e5"]
fig, axes = figure_grid(2, panel=(6.0, 4.0))
for ax, key, label in ((axes[0], "raan", "node Ω"), (axes[1], "argp", "argument of perihelion ω")):
    d = np.unwrap(fits[key][2]) * C.ARCSEC_PER_RAD
    ax.plot(t_yr, d - d[0], lw=0.8, color=PALETTE[1], label="(Newton + Lense–Thirring × 1e5) − Newton")
    p = float(rate_arcsec_per_century(pred[key]))
    ax.plot(t_yr, p * t_yr / 100.0, "k--", lw=1.0, label=f"Lense–Thirring: {p:+.2f}″/cy")
    ax.set_xlabel("t [years]"); ax.set_ylabel(f"Δ{label.split()[-1]} [arcsec]"); ax.grid(True, lw=0.3); ax.legend(fontsize=8)
    ax.set_title(f"{label}, solar-equator frame")
```

## 6. What a century of ranging can tell apart

Each effect leaves a different signature in the elements. GR touches only $\omega$ (the ellipse
turns in its own plane); the oblate Sun regresses the node and advances the argument, with
rates that depend on the inclination to the solar equator and fall off with distance faster
than GR's; frame dragging drags the node forward and the argument back; the EMB choice and the
asteroids are tiny, smooth and Newtonian. Everything below is from this lesson's ten-year runs;
the century-long versions of the solar terms ($+0.028\pm0.005$ and $-0.002\pm0.001''$/cy,
run-vs-run against Newton) are in `scripts/validate.py` and the `slow` tests of
`tests/test_extras.py` and `tests/test_moon.py`.

```python
j2_real_fits = j2_results["J2 = 2.2e-7 (real)"][1]
lt_real_fits = lt_results["J real"][1]
rows = (
    ("General Relativity", fits_gr["varpi"][0], fits_gr["raan"][0], fits_gr["argp"][0], "n/(c²p) ∝ a^-5/2 (1−e²)^-1; no i"),
    ("solar J2", j2_real_fits["varpi"][0], j2_real_fits["raan"][0], j2_real_fits["argp"][0], "n J2 (R/p)² ∝ a^-7/2 (1−e²)^-2; cos i, 5cos²i − 1"),
    ("Lense-Thirring", lt_real_fits["varpi"][0], lt_real_fits["raan"][0], lt_real_fits["argp"][0], "GJ/(c²a³(1−e²)^3/2); Ω free of i, ω ∝ cos i"),
    ("Moon vs EMB", shift[0], float("nan"), float("nan"), "a modelling choice: use the Moon"),
    ("asteroid belt", float(ring), float("nan"), float("nan"), "(3/4) n (m/M☉)(a/a_ring)³; not modelled"),
)
print("signature of each effect on Mercury, secular rates in arcsec/cy (solar-equator frame, ten-year runs)")
print(f"{'effect':20s} {'Δϖ':>10s} {'ΔΩ':>10s} {'Δω':>10s}   how it scales")
for name, dv, dO, dw, how in rows:
    print(f"{name:20s} {dv:+10.4f} {dO:+10.4f} {dw:+10.4f}   {how}")
residual = j2_real_fits["varpi"][0] + shift[0] + float(ring)
print(f"\nLesson 06's residual 'Horizons − EIH' was +0.025 arcsec/cy; the budget says J2 {j2_real_fits['varpi'][0]:+.4f}"
      f" + Moon {shift[0]:+.4f} + asteroids {float(ring):+.4f} = {residual:+.4f}: the ephemeris contains all three.")
```

## 7. Summary

* Mercury's budget: planets $531.6''$, GR $42.98''$, solar $J_2$ $+0.028''$, the Earth–Moon pair
  as one body $-0.0025''$, frame dragging $-0.002''$, asteroids $\sim+0.0014''$ per century,
  against a ranging precision of $\sim0.01''$. The solar $J_2$ is the only competitor that
  matters today, and Lesson 06's $0.025''$ residual against Horizons is exactly these terms.
* An oblate Sun regresses the node and advances the argument,
  $\dot\Omega = -\tfrac32 nJ_2(R/p)^2\cos i$, $\dot\omega = \tfrac34 nJ_2(R/p)^2(5\cos^2i-1)$;
  a spinning Sun drags the node forward, $2GJ/(c^2a^3(1-e^2)^{3/2})$, and the argument back three
  times faster. The runs reproduce both to $10^{-4}$, amplified or at the real size.
* From Mercury's perihelion alone $J_2$ and $2\gamma-\beta$ are degenerate; the node, the
  $1/p$ fall-off of the $J_2/{\rm GR}$ ratio across the planets, and helioseismology break it.
* The Moon as a body changes nothing for Mercury's GR signal, but the EMB point mass drifts
  from the real barycentre by $2\varepsilon nat \approx 110$ km/yr, $\varepsilon = \tfrac34 w_Ew_M(a_M/a_E)^2$.
  The lunar perigee advances $40.7°$ and the node regresses $19.3°$ per year — $25\,000$ times
  Mercury's advance, purely Newtonian, $20.1°$/yr at first order and doubled by Clairaut's
  second-order terms — and the ten-body run reproduces Horizons' rates over the same span to $0.01°$/yr.
